```python
#5.Write a Python program to check if a number is prime. Expected Output (for input 7):7 is a prime number
num = 7
fact = False
if num == 0 or num == 1:
    print(num, "is not a prime number")
elif num > 1: 
    for i in range(2, num):
        if (num % i) == 0:
            fact =   True
            break
    if fact == True:
        print(num, "is not a prime number")
    else:
        print(num, "is a prime number")
```
