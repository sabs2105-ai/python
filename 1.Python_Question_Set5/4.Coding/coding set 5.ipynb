```python
#1. Convert a List of Strings to UppercasebQuestion: Write a Python program to convert all strings in the list ["apple", "banana", "cherry"] to uppercase.

list =  ["apple", "banana", "cherry"]
for i in list:
    print(i.upper())
```

```python
#2. Merge Two Dictionaries Question: Write a Python program to merge two dictionaries:

dict1 = {'a': 10, 'b': 20}  
dict2 = {'c': 30, 'd': 40}

merged_dict = dict1 | dict2
print(merged_dict)
```

```python
#3. Find the Intersection of Two Lists Question: Write a Python program to find the common elements between the two lists:


list1 = [1, 2, 3, 4, 5]  
list2 = [4, 5, 6, 7, 8]
common = [x for x in list1 if x in list2]
print("Common Elements:", common)
```

```python
#4. Find the Length of Each Word in a Sentence Question: Write a Python program to print the length of each word in the sentence "Python is fun and powerful".

def print_word_lengths(sentence): 
    words = sentence.split()  
    for word in words:
        print(len(word))
sentence = "Python is fun and powerful"
print_word_lengths(sentence)
```

```python
#5. Find the Sum of All Elements in a 2D List Question: Write a Python program to find the sum of all elements in the 2D list:
def sum_2d_list(matrix):  
    total = 0
    for row in matrix:
        for item in row:
            if not isinstance(item, (int, float)):
                raise ValueError(f"Invalid element '{item}' - must be numeric.")
            total += item
    return total
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
result = sum_2d_list(matrix)
print(f"Sum of all elements in the matrix: {result}")
```
