```python
# Create a function that tells elegibility of marriage for male and female according to their age limit like 21 for male and 18 for female
```

```python
gender = input("Your Gender: ")
age = int(input("Your age: "))
def mrg_eligilility(gender,age):
    if(gender=="Male" and age >=21):
        return "ELIGIBLE"
    elif(gender=="Male" and age < 21):
        return "NOT ELIGIBLE"
    elif(gender=="Female" and age < 18):
        return "NOT ELIGIBLE"
    elif(gender=="Female" and age > 18 ):
        return "NOT ELIGIBLE"
eligibility = mrg_eligilility(gender,age)
print(eligibility)
```

```python
#print area and perimeter of triangle using functions
```

```python
Height=32
Breadth=34
def findarea(Height,Breadth):
    return (Height*Breadth)/2
area = findarea(Height,Breadth)
print("Area of Triangle:", area)

Height1=2
Height2=5
Breadth=7
def findperimeter(Height1,Height2,Breadth):
    return Height1+Height2+Breadth
per = findperimeter(Height1,Height2,Breadth)
print("Perimeter of Triangle: ", per)
```
