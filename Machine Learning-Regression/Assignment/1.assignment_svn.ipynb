```python
import pandas as pd
dataset=pd.read_csv("insurance_pre.csv")
```

```python
dataset=pd.get_dummies(dataset,drop_first=True)
dataset.columns
```

```python
independent=dataset[['age', 'bmi', 'children', 'sex_male', 'smoker_yes']]
dependent=dataset[["charges"]]
independent
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(independent, dependent, test_size=0.30,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler # here preprocessing is used to preprocess the train data to get better R2 value 
sc=StandardScaler() # This is preprocessor imported class.
X_train=sc.fit_transform(X_train) # Here need to transform train value to the decimal format
X_test=sc.transform(X_test)
X_train
```

```python
from sklearn.svm import SVR # support vecotr machine procedure
regressor=SVR(kernel="rbf",C=3000) # here kernal is using to find the better model Linear,rbf,poly,sigmond and C is hyper parameter
regressor.fit(X_train,y_train)
regressor.intercept_
regressor.n_support_ # This is to idenfy the total number of data inside the array
regressor.support_
y_pred=regressor.predict(X_test)
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
r_score
```

```python
import pickle
filename="finalized_model_Mul_linear.sav"
pickle.dump(regressor,open(filename,'wb'))
```

```python
loaded_model=pickle.load(open("finalized_model_Mul_linear.sav",'rb'))
result=loaded_model.predict([[1000,2000,3000,1,0]])
```
