```python
 #3.Find Even and Odd Numbers in a List uestion: Write a Python program to separate even and odd numbers from the list [12, 17, 19, 24, 35, 40, 56].
even = []
odd = []
def oddEven():
    nums = [12, 17, 19, 24, 35, 40, 56]
    for num in nums:
        if(num%2==0):
            even.append(num)
        else:
            odd.append(num)
    print(even)
    print(odd)
oddEven()
```
